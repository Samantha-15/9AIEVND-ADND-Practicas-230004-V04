```python
def bar_hor(df, col, title, color, w=None, h=None, lm=0, limit=100, return_trace=False, rev=False):
    cnt_srs = df[col].value_counts()
    yy = cnt_srs.head(limit).index[::-1] 
    xx = cnt_srs.head(limit).values[::-1] 
    if rev:
        yy = cnt_srs.tail(limit).index[::-1] 
        xx = cnt_srs.tail(limit).values[::-1] 
        
    
    trace = go.Bar(y=yy, x=xx, orientation = 'h', marker=dict(color=color))
    if return_trace:
        return trace 
    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    data = [trace]
    fig = go.Figure(data=data, layout=layout)
    iplot(fig)

def bar_hor_noagg(x, y, title, color, w=None, h=None, lm=0, limit=100, rt=False):
    trace = go.Bar(y=x, x=y, orientation = 'h', marker=dict(color=color))
    if rt:
        return trace
    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    data = [trace]
    fig = go.Figure(data=data, layout=layout)
    iplot(fig)


def bar_ver_noagg(x, y, title, color, w=None, h=None, lm=0, rt = False):
    trace = go.Bar(y=y, x=x, marker=dict(color=color))
    if rt:
        return trace
    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    data = [trace]
    fig = go.Figure(data=data, layout=layout)
    iplot(fig)

    
trace1 = bar_hor(donors_df, 'Donor City', "Top Cities with maximum Donors", '#c2d2ed', 500, 400, 200, limit = 10, return_trace=True)
trace2 = bar_hor(donors_df, 'Donor State', "Top States with maximum Donors", '#c2d2ed', 500, 400, 200, limit = 10, return_trace=True)
trace4 = bar_hor(donors_df, 'Donor State', "Top States with maximum Donors", '#c2d2ed', 600, 400, 200, limit = 10, return_trace=True, rev=True)


fig = tools.make_subplots(rows=1, cols=2, print_grid=False, subplot_titles = ['States with maximum Donors','States with minimum Donors'])
# fig = tools.make_subplots(rows=2, cols=2, specs=[[{}, {}], [{'colspan': 2}, None]], print_grid=False, subplot_titles = ['States with maximum Donors','States with minimum Donors', 'Cities with maximum Donors'])
fig.append_trace(trace2, 1, 1);
fig.append_trace(trace4, 1, 2);
# fig.append_trace(trace1, 2, 1);

fig['layout'].update(height=400, showlegend=False);
iplot(fig); 
```

### Datos Ajustado por Población

### 3.4 Donation Amounts Over Time

En esta sección se analiza la evolución de las donaciones a lo largo del tiempo para identificar tendencias en el comportamiento de los donantes. Se estudia tanto la cantidad de donaciones recibidas cada año como el monto promedio de las contribuciones, permitiendo observar cómo ha cambiado la actividad de la plataforma desde su creación.

Los siguientes análisis serán presentados:

- **3.4.1** Número de donaciones por año.
- **3.4.2** Monto promedio de las donaciones por año.

```python
# Convertir la columna de fecha
donations_df['Donation Received Date'] = pd.to_datetime(
    donations_df['Donation Received Date'],
    errors='coerce'
)

# Eliminar filas con fechas inválidas
donations_df = donations_df.dropna(subset=['Donation Received Date'])

# Crear variables de tiempo
donations_df['Donation Year'] = donations_df['Donation Received Date'].dt.year
donations_df['Donation Month'] = donations_df['Donation Received Date'].dt.month
donations_df['Donation Day'] = donations_df['Donation Received Date'].dt.day
donations_df['Donation Weekday'] = donations_df['Donation Received Date'].dt.weekday
donations_df['Donation Hour'] = donations_df['Donation Received Date'].dt.hour

# Agrupar por año
t1 = (
    donations_df.groupby('Donation Year')
    .agg(
        Total_Donations=('Donation ID', 'count'),
        Average_Amount=('Donation Amount', 'mean')
    )
    .reset_index()
)

# Crear gráficas
trace1 = go.Scatter(
    x=t1['Donation Year'],
    y=t1['Total_Donations'],
    fill='tozeroy',
    fillcolor='#fcc45f',
    mode='none'
)

trace2 = go.Scatter(
    x=t1['Donation Year'],
    y=t1['Average_Amount'],
    fill='tozeroy',
    fillcolor='#e993f9',
    mode='none'
)

fig = tools.make_subplots(
    rows=1,
    cols=2,
    subplot_titles=(
        "<b>Donations per Year</b>",
        "<b>Average Donation Amount per Year</b>"
    ),
    print_grid=False
)

fig.append_trace(trace1, 1, 1)
fig.append_trace(trace2, 1, 2)

fig.update_layout(
    height=350,
    width=900,
    showlegend=False,
    template='plotly_white'
)

fig.update_yaxes(
    showgrid=False,
    zeroline=False,
    showticklabels=False,
    row=1,
    col=1
)

fig.update_yaxes(
    showgrid=False,
    zeroline=False,
    showticklabels=False,
    row=1,
    col=2
)

iplot(fig)
```

Desde el año **2012**, el número de donaciones en **DonorsChoose** ha aumentado de manera constante. Mientras que en 2012 se registraron alrededor de **150 donaciones**, en **2017** la plataforma recibió aproximadamente **1.1 millones de donaciones**.

- El **monto de donación más alto** recibido en **2017** fue de **USD 31,856**, realizado por un donante de **Lafayette, Colorado**. Esta donación se registró el **2 de marzo de 2017**.

- En ese mismo año, un donante de **Manhattan Beach, California** realizó un total de **13,751 donaciones**, siendo uno de los mayores contribuyentes en cuanto al número de donaciones.

- El **monto promedio de donación por año** se mantiene cercano a **USD 60**. Sin embargo, el promedio más alto se registró en el año inicial del conjunto de datos (**2012**), con un valor aproximado de **USD 166**.

```python
def bar_ver_noagg(x, y, title, color, w=None, h=None, lm=60, rt=False):

    trace = go.Bar(
        x=x,
        y=y,
        marker=dict(color=color)
    )

    if rt:
        return trace

    layout = go.Layout(
        title=dict(
            text=title,
            x=0.5
        ),

        width=w,
        height=h,

        plot_bgcolor='white',
        paper_bgcolor='white',

        margin=dict(
            l=lm,
            r=20,
            t=70,
            b=50
        ),

        showlegend=False,

        xaxis=dict(
            showgrid=False,
            showline=False,
            zeroline=False
        ),

        yaxis=dict(
            showgrid=True,
            gridcolor='#EAEAEA',
            zeroline=False
        )
    )

    fig = go.Figure(data=[trace], layout=layout)

    iplot(fig)
```

Se define una función auxiliar para **normalizar el texto y eliminar el ruido**. Esta función de limpieza realiza la eliminación de **palabras vacías (*stopwords*)** y de los **signos de puntuación** presentes en el texto.

```python
# ============================
# Librerías
# ============================
import nltk
import string
from nltk.corpus import stopwords
from textblob import TextBlob

# ============================
# Descargar recursos de NLTK
# (solo la primera vez)
# ============================

try:
    nltk.data.find('corpora/stopwords')
except LookupError:
    nltk.download('stopwords')

# ============================
# Lista de stopwords
# ============================

stop_words = set(stopwords.words("english"))

# ============================
# Función de limpieza
# ============================

def _clean(txt):

    # Evitar errores con valores nulos
    if pd.isna(txt):
        return ""

    txt = str(txt).lower()

    # Eliminar signos de puntuación
    txt = "".join(c for c in txt if c not in string.punctuation)

    clean_txt = []

    for word in txt.split():
        if word not in stop_words:
            clean_txt.append(word)

    clean_txt = " ".join(clean_txt)

    # Palabras consideradas ruido
    noise = [
        'students',
        'student',
        'my',
        'title',
        'would',
        'every',
        'will',
        'many',
        'also',
        'donotremoveessaydivider'
    ]

    for n in noise:
        clean_txt = clean_txt.replace(n, "")

    return clean_txt.strip()

# ============================
# Unir datasets
# ============================

projsch_df = projects_df.merge(
    schools_df,
    on="School ID",
    how="inner"
)

# ============================
# Filtrar proyectos de Abril 2018
# ============================

april18_df = projsch_df[
    (projsch_df["Posted Year"] == 2018) &
    (projsch_df["Posted Month"] == 4)
].copy()

# ============================
# Limpiar ensayos
# ============================

april18_df["clean_essay"] = (
    april18_df["Project Essay"]
    .fillna("")
    .apply(_clean)
)

# ============================
# Limpiar Need Statement
# ============================

if "Project Need Statement" in april18_df.columns:

    april18_df["clean_need"] = (
        april18_df["Project Need Statement"]
        .fillna("")
        .apply(_clean)
    )

else:

    print("La columna 'Project Need Statement' no existe en este dataset.")
    april18_df["clean_need"] = ""
```

```python
### Bag of Words 
from collections import Counter
def generate_ngrams(txt, N):
    grams = [txt[i:i+N] for i in range(len(txt)-N+1)]
    grams = [" ".join(b) for b in grams]
    return grams 

april18_df['unigrams'] = april18_df['clean_essay'].apply(lambda x : generate_ngrams(x.split(), 1))
april18_df['bigrams'] = april18_df['clean_essay'].apply(lambda x : generate_ngrams(x.split(), 2))
april18_df['trigrams'] = april18_df['clean_essay'].apply(lambda x : generate_ngrams(x.split(), 3))

all_unigrams = []
for each in april18_df['unigrams']:
    all_unigrams.extend(each)
t = Counter(all_unigrams).most_common(25)
x = [a[0] for a in t]
y = [a[1] for a in t]

all_bigrams = []
for each in april18_df['bigrams']:
    all_bigrams.extend(each)
t1 = Counter(all_bigrams).most_common(25)
x1 = [a[0] for a in t1]
y1 = [a[1] for a in t1]


all_trigrams = []
for each in april18_df['trigrams']:
    all_trigrams.extend(each)
t2 = Counter(all_trigrams).most_common(25)
x2 = [a[0] for a in t2]
y2 = [a[1] for a in t2]

sns.set_style("dark")
fig, axes = plt.subplots(nrows=1, ncols=5, squeeze=True, figsize = (12,10));

bar = sns.barplot(y=x, x=y, ax=axes[0], palette='GnBu_d', edgecolor="white");
bar.set(xlabel='', xticks=[]);
axes[0].set_title("Top Keywords used by Teachers");

fig.delaxes(axes[1]);

bar1 = sns.barplot(y=x1, x=y1, ax=axes[2], palette='GnBu_d', edgecolor="white");
bar1.set(xlabel='', xticks=[]);
axes[2].set_title("Top Bigrams used by Teachers");

fig.delaxes(axes[3]);

bar1 = sns.barplot(y=x2, x=y2, ax=axes[4], palette='GnBu_d', edgecolor="white");
bar1.set(xlabel='', xticks=[]);
axes[4].set_title("Top Trigrams used by Teachers");
```
